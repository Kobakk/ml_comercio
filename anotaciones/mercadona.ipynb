```python
import kagglehub
from kagglehub import KaggleDatasetAdapter
import pandas as pd
from pathlib import Path
import os
from dotenv import load_dotenv
import shutil
```

### Importamos los datos

#### posibles errores:

- Conflicto con cache: Las APIs como kagglehub están diseñadas para ser eficientes.
- Conflicto con rutas: En entornos como Jupyter Lab dentro de Pycharm pude ser traicionero, tenemos que usar rutas absolutas.
- Sintaxís del archivo `.env`

#### código

```python
load_dotenv()
print(os.getenv("KAGGLE_USERNAME"))
print(os.getenv("KAGGLE_KEY"))
```

```python
ruta_origen = kagglehub.dataset_download("computingvictor/mercadoan-inventory")
print(ruta_origen)
```

```python
# 1. Definimos rutas
# Usamos Path para que funcione bien tanto en Windows como en Linux
ruta_destino = os.path.abspath("../datasets")
os.makedirs(ruta_destino, exist_ok=True)

# 2. Verificamos qué hay en la fuente (la caché de kagglehub)
archivos_fuente = os.listdir(ruta_origen)

if not archivos_fuente:
    print(f"❌ Error: No hay archivos en {ruta_origen}")
    print("💡 Probablemente los moviste antes. Borra la caché y descarga de nuevo.")
else:
    print(f"📦 Encontrados {len(archivos_fuente)} archivos. Iniciando transferencia...")

    for archivo in archivos_fuente:
        punto_origen = os.path.join(ruta_origen, archivo)
        punto_destino = os.path.join(ruta_destino, archivo)

        # 3. Usamos copy2 en lugar de move
        # copy2 preserva fechas y metadatos, y NO borra el origen.
        shutil.copy2(punto_origen, punto_destino)
        print(f"  ✅ Copiado: {archivo}")

    print(f"\n🚀 Proceso terminado. Tus archivos están en: {ruta_destino}")
```
