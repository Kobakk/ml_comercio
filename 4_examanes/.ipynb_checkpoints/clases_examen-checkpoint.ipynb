Este cuaderno permite crear un sistema en Python para **gestionar y analizar datasets** con características **numéricas** y **categóricas**.

## ¿Qué es un dataset?

Un **dataset** (o conjunto de datos) es una colección organizada de información que se utiliza para análisis, aprendizaje automático o toma de decisiones. Cada **muestra** del dataset representa un caso o registro, y cada **columna** representa una característica o atributo de ese caso.

* **Muestras (filas)**: cada fila contiene los datos de un ejemplo individual.
* **Características (columnas)**: cada columna contiene un atributo de los ejemplos.
* **Tipos de características**:

  * **Numéricas**: valores que se pueden medir y usar en cálculos matemáticos (ej. edad, altura, peso).
  * **Categóricas**: valores que representan categorías o etiquetas (ej. color, forma, tipo de material).

### Ejemplo de dataset

Supongamos que tenemos un dataset de personas:

| nombre | edad | altura | color_ojos |
| ------ | ---- | ------ | ---------- |
| Ana    | 25   | 1.70   | azul       |
| Juan   | 30   | 1.82   | marrón     |
| María  | 22   | 1.65   | verde      |

* **Filas**: cada persona es una muestra.
* **Columnas**:

  * `edad` y `altura` → numéricas
  * `nombre` y `color_ojos` → categóricas

Este cuaderno permitirá:

1. **Crear datasets** como este.
2. **Añadir, modificar o eliminar muestras** fácilmente.
3. **Guardar y cargar** datasets en formato JSON.
4. **Realizar análisis básicos**:

   * Calcular la **media** de columnas numéricas (`edad`, `altura`)
   * Contar la frecuencia de cada categoría en columnas categóricas (`color_ojos`)

## 🧩 **Clase Base: `Dataset`**

### 📘 Propósito

Sirve como **clase abstracta** para definir la estructura y comportamiento general de un dataset (colección de muestras).
No debe instanciarse directamente, sino a través de subclases concretas.

### 📋 Atributos

* `__nombre`: nombre identificativo del dataset (string).
* `__datos`: lista de muestras, donde cada muestra es un **diccionario** con los datos de una fila.

### 🧱 Propiedades

* `nombre`: solo lectura. Devuelve el nombre del dataset.
* `datos`: permite acceder y asignar los datos del dataset.

  * Debe ser una **lista de diccionarios**.
  * Si se intenta asignar otro tipo, lanza `TypeError`.

### 🔧 Métodos abstractos

* `cargar_fichero(nombre_fichero)`: debe cargar el dataset desde un archivo (normalmente JSON).
* `guardar_fichero(nombre_fichero)`: debe guardar el dataset en un archivo.

### 🧠 Métodos concretos

* `añadir_muestra(muestra)`:
  Agrega una muestra al dataset.

  * Solo acepta diccionarios.
  * Si se pasa otro tipo, lanza `TypeError`.

* `eliminar_muestra(indice)`:
  Elimina la muestra en la posición indicada.

  * Si el índice no existe, lanza `IndexError`.

* `obtener_muestra(indice)`:
  Devuelve la muestra en la posición indicada.

  * Si el índice no existe, lanza `IndexError`.





```python
# TODO: Dataset
from abc import ABC, abstractmethod

class Dataset(ABC):
    def __init__(self, nombre):
        self.__nombre = nombre
        self.__datos = []

    @property
    def datos(self):
      return self.datos

    @property
    def nombre(self):
      return self.__nombre

    @datos.setter
    def datos(self, datos):
      self.__datos = datos
    # metodos abstractor
    @abstractmethod
    def cargar_fichero(self, n_fichero):
      pass
    @abstractmethod
    def guardar_fichero(self, n_fichero):
      pass

    def nombre(self):
      return self.__nombre

    def eliminar_muestra(self, indice):
      try:
        del self.__datos[indice]
      except IndexError as e:
        raise e("el indice no existe")

    def anadir_muestra(self, dato):
      self.__datos.append(dato)
    def obtener_muestra(self, indice):
      return self.__datos[indice]
```

## 🔢 **Clase `DatasetNumérico` (hereda de `Dataset`)**

### 📘 Propósito

Gestiona datasets cuyos valores son **numéricos** (int o float).

### ⚙️ Métodos sobrescritos y añadidos

* `añadir_muestra(muestra)`:

  * Puede recibir:

    * Un **diccionario** (comportamiento estándar).
    * Una **tupla**: en este caso convierte la tupla a diccionario tomando las claves del primer elemento ya existente en `datos`.
  * Valida:

    * Que exista al menos una muestra previa (para conocer las columnas).
    * Que la tupla tenga el mismo número de elementos que las columnas.
  * Si algo falla, lanza `ValueError`.

* `media(columna)`:

  * Calcula la media aritmética de los valores de una columna.
  * Valida que:

    * La columna exista (`KeyError`).
    * Todos los valores sean numéricos (`TypeError`).
  * Devuelve `0` si no hay datos.

* `cargar_fichero(nombre_fichero)`:

  * Abre y carga un fichero JSON.
  * Valida que el fichero exista (`FileNotFoundError`).

* `guardar_fichero(nombre_fichero)`:

  * Guarda los datos actuales como JSON.
  * Si ocurre un error al escribir, lanza `IOError`.



```python
 # TODO: DatasetNumerico
import json
class DatasetNumerico(Dataset):

  def anadir_muestra(self, muestra):
    clave_dataset = list(self.datos[0].keys())
    muestra = dict(zip(clave_dataset, muestra))
    super.anadir_muestra(muestra)

def media(self, columna):
  if not self.datos:
    return 0
  valor = [dato[columna] for dato in self.datos]
  return sum(valor) /len(valor)

def cargar_fichero(self, nombre_fichero):
  with open(nombre_fichero, "r") as f:
    data = json.load(f)
    self.datos = data


def guardar_fichero(self, nombre_fichero):
  with open(nombre_fichero, "w") as f:
    json.dump(self.datos, f, indent=2)

```

## 🔤 **Clase `DatasetCategorico` (hereda de `Dataset`)**

### 📘 Propósito

Gestiona datasets cuyos valores son **categóricos** (texto, etiquetas, etc.).

### ⚙️ Métodos añadidos

* `conteo_categoria(columna)`:

  * Devuelve un diccionario con la **frecuencia de cada categoría** en la columna indicada.
  * Si la columna no existe, lanza `KeyError`.

* `cargar_fichero(nombre_fichero)` y `guardar_fichero(nombre_fichero)`:

  * Funcionan igual que en `DatasetNumérico`.
  * Gestionan archivos JSON con los datos categóricos.



```python
# TODO: DatasetCategorico
class DatasetCategorico(Dataset):

  def conteo_categoria(self, columna):
    conteo = 0
    for muestra in self.datos():

      valor = muestra[columna]
      #print(valor)
      conteo[valor]= conteo.get(valor, 0) + 1
    return conteo
  def cargar_fichero(self, nombre_fichero):
    with open(nombre_fichero, "r") as f:
      data = json.load(f)
      self.datos = data
  def guardar_fichero(self, nombre_fichero):
    with open(nombre_fichero, "w") as f:
      json.dump(self.datos, f, indent=2)
```
